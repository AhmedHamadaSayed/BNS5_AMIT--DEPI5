```python
import pandas as pd


def automated_stat_analyzer(df, column_name):
    """
    Analyze a column and return its main statistics.
    """
    if pd.api.types.is_numeric_dtype(df[column_name]):

        mean = df[column_name].mean()
        median = df[column_name].median()
        std = df[column_name].std()

        
        if mean > median:
            skewed = "Right Skewed"
        elif mean < median:
            skewed = "Left Skewed"
        else:
            skewed = "Symmetric"

        return {
            "Mean": mean,
            "Median": median,
            "Standard Deviation": std,
            "Skewness": skewed
        }

    
    else:
        mode = df[column_name].mode()[0]

        return {
            "Mode": mode
        }



result = automated_stat_analyzer(df_test, "Sales_Amount")
print(result)
print("#" * 50)

result2 = automated_stat_analyzer(df_test, "Product_Category")
print(result2)
```

```python
import pandas as pd


def null_handling_strategy(df, strategy="fill_mean"):
    """
    Handle missing values using the selected strategy.
    """

    
    if not df.isnull().values.any():
        return df

    
    if strategy == "drop_rows":
        return df.dropna()

    
    elif strategy == "fill_mean":
        numerical_cols = df.select_dtypes(include="number").columns

        for col in numerical_cols:
            df[col] = df[col].fillna(df[col].mean())

        return df

    
    elif strategy == "fill_median":
        numerical_cols = df.select_dtypes(include="number").columns

        for col in numerical_cols:
            df[col] = df[col].fillna(df[col].median())

        return df

    else:
        return "Invalid strategy"
```
