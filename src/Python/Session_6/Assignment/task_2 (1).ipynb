1. Two Sum
EasyTopicsCompanies
You are given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.You may assume that each input would have exactly one solution, and you may not use the same element twice.You can return the answer in any order. Example 1:Input: nums = [2,7,11,15], target = 9
Output: [0,1]

```python
class Two_sum:
    '''
    Solve the two sum problem

    Attributes :
    target : int 
    The target sum 

    num :list[int]
    list of integers

    '''
    def __init__(self,target,num):
        '''Store the target and num '''
        self.target = target 
        self.num = num 

    def two_sum(self):
        '''Find two indices whose values add up to the target'''
        for i in range (len(self.num)):
            for j in range(i+1 ,len(self.num)):
                if self.num[i]+self.num[j] == self.target :
                    return[i,j]
result = Two_sum(9,[2,7,11,15])
result.two_sum()
```

```python
class ListNode:
    '''A node in the linked list.
    It stores a value and a link to the next node.
    '''
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
        '''
        Adds two numbers stored in linked lists.
        Returns a new linked list with the result.
        '''
    def addTwoNumbers(self, l1, l2):

        dummy = ListNode(0)
        current = dummy
        carry = 0

        while l1 or l2 or carry:

            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            total = val1 + val2 + carry

            digit = total % 10
            carry = total // 10

            current.next = ListNode(digit)
            current = current.next

            if l1:
                l1 = l1.next

            if l2:
                l2 = l2.next

        return dummy.next


def create_linked_list(values):
    '''Creates a linked list from a list of values.'''
    head = ListNode(values[0])
    current = head

    for value in values[1:]:
        current.next = ListNode(value)
        current = current.next

    return head


def print_linked_list(head):
    '''
    Prints the linked list values as a normal list.
    '''
    values = []

    while head:
        values.append(head.val)
        head = head.next

    print(values)


l1 = create_linked_list([2, 4, 3])
l2 = create_linked_list([5, 6, 4])

solution = Solution()

result = solution.addTwoNumbers(l1, l2)

print_linked_list(result)
```
