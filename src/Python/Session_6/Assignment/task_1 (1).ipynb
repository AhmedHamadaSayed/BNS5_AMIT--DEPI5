Write a Class that inputs a number and prints the multiplication table of that 
number 

```python
class Multi_table:
    '''
    This Class Print Multiplication Table of a number

    Attribute:
    num : int
    
    '''
    def __init__(self,num):
        '''Store the num '''
        
        self.num = num 

    def multiplication(self):
        ''' Print The Multiplication table from 1 to 10'''

        for i in range (1,11):
            print (f"{self.num} x {i} = {self.num * i}")    

table = Multi_table(5)  
table.multiplication()          
```

 Write a class to print twin primes less than 1000. If two 
consecutive odd numbers are both prime then they are known as 
twin primes

```python
class Twin_primes:
    ''' this class print the twin prime number
    Attribute:
    limit : int 
    the maximum number
    
    '''
    def __init__(self,limit):
        '''
        Store the limit
        
        '''
        self.limit = limit 

    def is_prime(self,num)  :
        '''Check The Number Is Prime 
        
        returns:
        bool : True if the number is prime
        
        '''  
        self.num = num 

        if num < 2:
            return False
        for i in range (2,num):
            if num % i == 0 :
                return False
        return True 

    def  print_twin_prime(self):
        ''' print the twin prim less than limit ''' 

        for i in range (2,self.limit):
            if self.is_prime(i) and self.is_prime(i+2):
                print(i , i+2)
twin_primes = Twin_primes(1000)  
twin_primes.print_twin_prime()              
```

 Write a class to find out the prime factors of a number. 
Example: prime factors of 56 - 2, 2, 2, 7

```python
class Prime_factors:
    '''This Class Find The prime factors of number 
    Attribute:
    num : int 
    The number to find its prime factors
    
    '''
    def __init__(self,num):
        '''Store the num '''
        self.num = num 

    def print_prime_factors(self):
        '''Print the prime factors of the number '''   

        factors=[]

        for i in range (2,self.num+1):
            while self.num % i == 0 :
                factors.append(i)
                self.num = self.num // i

        return factors        
prime_factors = Prime_factors(56)
prime_factors.print_prime_factors()
```

Write a function that converts a decimal number to binary 
number 

```python
class Decimal_to_binary:
    '''This Class convert  a decimal number to binary 

    Attribute:
    num = int 
    this number which convert to binary 
    '''
    def __init__ (self,num):
        '''Store the num '''
        
        self.num = num 
    def convert (self):
        '''Convert the decimal number to binary.

        Returns:
            str: The binary number'''
        Binary = ''
        
        while self.num >0:
            remainder = self.num % 2
            Binary += str(remainder)
            self.num = self.num // 2
        return Binary
decimal_number = Decimal_to_binary(10)  
reselt = decimal_number.convert()
print(reselt)
```

A number is called perfect if the sum of proper divisors of that 
number is equal to the number. For example 28 is perfect 
number, since 1+2+4+7+14=28. Write a program to print all the 
perfect numbers in a given range

```python
class Perfect_Numbers:
    '''This Class finds perfect numbers in a given range
    Attributes:
    start : int 

    The start of the range

    end : int

    The end of the range 

    '''
    def __init__(self,start,end):
        '''Stor the range'''
        self.start = start
        self.end = end 

    def is_perfect(self,num):
        '''Check if number is perfect
        returns:
        bool: True if the number is perfect
        
        '''
        total = 0
        for i in range (1,num):
            if num % i == 0:
                total += i
        return total == num

    def perfect_numbers(self):
        '''Print the all perfect number in range''' 

        for num in range (self.start,self.end +1):
            if self.is_perfect(num)  :
                print(num)


perfect_numbers = Perfect_Numbers(1,1000)
perfect_numbers.perfect_numbers()
               
```

Objective: Create a simple calculator class that can perform basic arithmetic operations: addition, subtraction, multiplication, and division.

"*"Requirements:"*"
------------------------
"*"User Input:"*"

The program should prompt the user to choose an operation from the following list:
- Addition
- Subtraction
- Multiplication
- Division
- Exit
Input Validation:

Ensure that the user inputs valid numbers for the calculations. If the input is invalid (e.g., non-numeric), prompt the user to enter the numbers again.
Performing Calculations:

Based on the user's choice, the program should ask for two numbers and perform the selected operation.
If the user selects division, check if the second number is zero to avoid division by zero errors.
Looping:

After each calculation, ask the user if they want to perform another calculation. If they answer "yes," return to the operation selection; if they answer "no," exit the program.
Output:

Display the result of the calculation in a user-friendly format.
For example: "The result of adding 5 and 3 is 8."
Exit Option:

If the user chooses to exit, print a thank you message and terminate the program.

```python
class Calclator:
    '''This class performs basic arithmetic operations.

    Attributes:
        num1 : float
            The first number.
        num2 : float
            The second number.
    '''

    def __init__(self):
        '''Initialize the calculator'''
        self.num1 = 0
        self.num2 = 0

    def calc(self):
        '''Perform the selected arithmetic operation.

        Returns:
            float: The result of the calculation.
        '''

        while True:
            print("1. Addition")
            print("2. Subtraction")
            print("3. Multiplication")
            print("4. Division")
            print("5. Exit")

            choice = int(input("Enter your choice: "))

            if choice == 5:
                print("Thank you for using the calculator!")
                break

            self.num1 = float(input("Enter first number: "))
            self.num2 = float(input("Enter second number: "))

            if choice == 1:
                result = self.num1 + self.num2
                print(f"The result of adding {self.num1} and {self.num2} is {result}")

            elif choice == 2:
                result = self.num1 - self.num2
                print(f"The result of subtracting {self.num2} from {self.num1} is {result}")

            elif choice == 3:
                result = self.num1 * self.num2
                print(f"The result of multiplying {self.num1} and {self.num2} is {result}")

            elif choice == 4:
                if self.num2 == 0:
                    print("Cannot divide by zero")
                else:
                    result = self.num1 / self.num2
                    print(f"The result of dividing {self.num1} by {self.num2} is {result}")

            else:
                print("Invalid choice")

            again = input("Do you want to perform another calculation? ")

            if again.lower() != "yes":
                print("Thank you for using the calculator!")
                break
calclator = Calclator()
calclator.calc()
```

```python
import os
import random


class ThanosProject:
    '''
    This class creates files and randomly deletes half of them

    Attributes:
        folder_path (str): The path of the folder
        number_of_files (int): The number of files to create
    '''

    def __init__(self, folder_path, number_of_files):
        '''
        Store the folder path and number of files

        '''
        self.folder_path = folder_path
        self.number_of_files = number_of_files

    def create_files(self):
        """
        Create the required number of files inside the folder.
        """

        if not os.path.exists(self.folder_path):
            os.makedirs(self.folder_path)

        for i in range(1, self.number_of_files + 1):
            file_path = os.path.join(
                self.folder_path,
                f"file_{i}.txt"
            )

            with open(file_path, "w") as file:
                file.write(f"This is file {i}")

    def get_files(self):
        """
        Get all files inside the folder.

        Returns:
            list: A list of files.
        """

        files = [
            file for file in os.listdir(self.folder_path)
            if os.path.isfile(
                os.path.join(self.folder_path, file)
            )
        ]

        return files

    def delete_half(self):
        """
        Delete a random half of the files.
        """

        files = self.get_files()

        number_to_delete = len(files) // 2

        files_to_delete = random.sample(
            files,
            number_to_delete
        )

        for file in files_to_delete:
            file_path = os.path.join(
                self.folder_path,
                file
            )

            os.remove(file_path)

    def run(self):
        """
        Run the complete Thanos project.
        """

        self.create_files()

        files_before = self.get_files()

        print(
            f"Number of files before Thanos: "
            f"{len(files_before)}"
        )

        self.delete_half()

        files_after = self.get_files()

        print(
            f"Number of files after Thanos: "
            f"{len(files_after)}"
        )


thanos_project = ThanosProject("thanos_folder", 10)

thanos_project.run()
```
