1. Write a function that inputs a number and prints the multiplication table of that 
number


```python
def multiplication (num :int):
    for i in range(1,11):
        print(f"{num} x {i} = {num * i}")
```

```python
number = int (input("Enter your number:"))
multiplication(number)
```

2. Write a program to print twin primes less than 1000. If two 
consecutive odd numbers are both prime then they are known as

```python
def is_prime(num:int)->bool:
    '''
     this function check the number is prime 

     parameters:
     -----------
     num:int

     return:
     -------
     bool
    '''
    if num < 2:
        return False
    for i in range(2,num):  
        if num % i ==0:
            return False
    return True       
```

```python
for number in range (2,1000):
    if is_prime (number ):
        if is_prime (number+2):
            print(number,number+2)
```

4. Write a function that converts a decimal number to binary 
number 

```python
def decimal_to_binary(num):
    Binary = " "
    while num>0:
        remainder = num%2
        Binary +=str(remainder)
        num =num //2
    return Binary    
```

 5.A number is called perfect if the sum of proper divisors of that 
number is equal to the number. For example 28 is perfect 
number, since 1+2+4+7+14=28. Write a program to print all the 
perfect numbers in a given range

```python
start = int(input("Enter the start of the range:"))
end = int(input("Enter the end of the range:"))
for i in range (start ,end +1):
    total = 0
    for divisors in range (1,i):
        if i % divisors ==0:
            total+=divisors
    if total == i:
       print(total)        
```
